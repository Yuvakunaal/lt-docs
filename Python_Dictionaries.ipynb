## 1. What is a Dictionary?

A **dictionary** stores data in **key : value** pairs. You find data using a meaningful **key**, not a number position.

**Real life example:** Think of a **real dictionary book**. You look up a **word** (key) and get its **meaning** (value) — you don't search by page number.

**Symbol used:** Curly braces `{ }` with `key: value` pairs.

## 2. Creating a Dictionary

````python
student = {"name": "Arun", "age": 21, "course": "Python"}
print(student)
````

## 3. Accessing Values

Two ways: using `[ ]` or using `.get()`.

- `student["marks"]` → gives an **error** if key doesn't exist
- `student.get("marks", "Not Found")` → gives a **safe default**, no error

````python
print(student["name"])
print(student.get("age"))
print(student.get("marks", "Not Found"))
````

## 4. Adding and Updating Values

Assign a value to a key. New key → added. Existing key → updated.

````python
student["marks"] = 85
print(student)

student["age"] = 22
print(student)
````

## 5. Removing Items from a Dictionary

| Method | What it does |
|---|---|
| `pop(key)` | Removes a specific key and returns its value |
| `popitem()` | Removes the last inserted key-value pair |
| `del dict[key]` | Deletes a specific key |
| `clear()` | Empties the whole dictionary |

````python
student.pop("marks")
print(student)

d2 = student.copy()
del d2["course"]
print(d2)

last = student.popitem()
print("removed:", last)
print(student)
````

## 6. Looping Through a Dictionary

- Looping directly → gives **keys**
- `.values()` → gives only **values**
- `.items()` → gives **both** key and value together (most useful)

````python
person = {"name": "Divya", "city": "Chennai", "job": "Engineer"}

for key in person:
    print(key)

print("--- values ---")
for value in person.values():
    print(value)

print("--- items ---")
for key, value in person.items():
    print(key, "->", value)
````

## 7. Important Methods: `keys()`, `values()`, `items()`

`.items()` gives a list of **tuples** — each tuple has one key and one value together.

````python
print(list(person.keys()))
print(list(person.values()))
print(list(person.items()))
````

## 8. `update()` Method

Used to add or change multiple values at once.

````python
person.update({"age": 23})
print(person)
````

## 9. `setdefault()` Method

- If the key **doesn't exist** → adds it with the given value
- If the key **already exists** → does nothing, returns existing value

````python
person.setdefault("country", "India")
print(person)

person.setdefault("city", "Madurai")
print(person)
````

Notice above: `"city"` did **not** change to "Madurai" — because `"city"` already had value `"Chennai"`.

## 10. `fromkeys()` Method

Quickly creates a new dictionary from a list of keys, all with the same starting value.

**Real life example:** Like giving every student in a class a starting mark of "0" before results come in.

````python
keys = ["a", "b", "c"]
new_dict = dict.fromkeys(keys, 0)
print(new_dict)
````

## 11. Checking if a Key Exists

Use the `in` keyword — very useful to avoid errors before accessing a key.

````python
print("name" in person)
print("salary" in person)
````

## 12. Nested Dictionary (Dictionary Inside a Dictionary)

**Real life example:** Like a school register — the register (outer dictionary) has roll numbers, and each roll number has a small box (inner dictionary) with that student's own details.

````python
students = {
    "s1": {"name": "Ravi", "marks": 80},
    "s2": {"name": "Kumar", "marks": 90}
}

print(students["s1"]["name"])

for sid, info in students.items():
    print(sid, info["name"], info["marks"])
````

## 13. Dictionary Comprehension

A short, one-line way of creating a dictionary using a loop.

```python
{key_expression: value_expression for item in iterable}
```

````python
squares = {x: x*x for x in range(1, 6)}
print(squares)
````

## 14. Merging Two Dictionaries

Use `**` unpacking inside `{ }`. If a key exists in both, the **later** dictionary's value wins.

````python
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
merged = {**d1, **d2}
print(merged)
````

## 15. `len()` and `clear()`

`len()` tells how many key-value pairs exist. `clear()` empties the dictionary completely.

````python
print(len(merged))

temp = merged.copy()
temp.clear()
print(temp)
````
