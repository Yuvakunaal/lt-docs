## 1. What is a Function?

A **function** is a block of code that does one particular job. You write it once and use it (call it) again and again.

**Real life example:** A function is like a **mixer grinder**. You put vegetables in (input), press the button (call the function), and get juice out (output). You don't build a new mixer every time.

**Why use functions?**
- Avoid repeating code
- Keep code clean and readable
- Easy to fix (change once, works everywhere)
- Break a big problem into small pieces

## 2. Creating (Defining) a Function

We use the keyword `def` to create a function.

```python
def function_name():
    # code to run
```

````python
def greet():
    print("Hello, welcome to Python!")

greet()   # calling the function
````

👉 `def greet():` **creates** the function.
👉 `greet()` **calls / runs** it. If you don't call it, it never runs.

## 3. Function with Parameters

A **parameter** is a gap you leave in your function so people can give you information to work with.

````python
def greet_person(name):
    print("Hello, " + name + "!")

greet_person("Ravi")
````

**Difference:**
- **Parameter** = the empty box in the function definition
- **Argument** = the actual item you put inside that box when calling

## 4. Function with Return Value

Use `return` to send a value back so you can store and reuse it, instead of just printing it.

````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
````

## 5. Default Arguments

A parameter can have a default value, used only when the caller doesn't provide one.

**Real life example:** If you order coffee and don't say "less sugar", the shop gives the **default** amount.

````python
def greet_default(name="Friend"):
    print("Hi " + name)

greet_default()
greet_default("Anitha")
````

## 6. Keyword Arguments

Arguments can be passed using the parameter's name — order doesn't matter then.

````python
def student_info(name, age):
    print(name, "is", age, "years old")

student_info(age=20, name="Kumar")
````

## 7. Positional vs Keyword Arguments Together

````python
def order(item, qty, price):
    print(f"{qty} x {item} = {qty*price}")

order("Pen", 3, 10)
order(item="Book", price=50, qty=2)
````

## 8. `*args` — Unlimited Positional Arguments

`*args` lets a function accept **any number** of values.

**Real life example:** A **school bus** (`*args`) can hold 3 students or 50 students — it doesn't matter.

````python
def add_all(*numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(add_all(1, 2, 3))
print(add_all(1, 2, 3, 4, 5))
````

## 9. `**kwargs` — Unlimited Keyword Arguments

`**kwargs` accepts any number of `key=value` pairs. Inside the function, it becomes a dictionary.

````python
def print_info(**details):
    for key, value in details.items():
        print(key, ":", value)

print_info(name="Sam", age=25, city="Chennai")
````

## 10. Returning Multiple Values

A function can return more than one value at once, separated by commas (Python packs them into a tuple).

````python
def calc(a, b):
    return a+b, a-b, a*b, a/b

s, d, m, dv = calc(10, 5)
print(s, d, m, dv)
````

## 11. Docstrings

A **docstring** is a short description written in triple quotes right after the function definition.

````python
def square(n):
    """This function returns square of a number"""
    return n*n

print(square(6))
print(square.__doc__)
````

## 12. Variable Scope — Local vs Global

- **Local variable:** exists only inside the function
- **Global variable:** exists everywhere in the program

**Real life example:** "x" is like a nickname. Inside your house (function) it can mean one person, outside on your street (global) it can mean someone else — they don't clash.

````python
x = 10

def show():
    x = 5
    print("Inside function x =", x)

show()
print("Outside function x =", x)
````

## 13. The `global` Keyword

Use `global` if you actually want to change the outer variable from inside a function.

````python
count = 0

def increment():
    global count
    count += 1

increment()
increment()
print(count)
````

## 14. Nested Functions (Function Inside a Function)

````python
def outer():
    print("Outer function")
    def inner():
        print("Inner function")
    inner()

outer()
````

## 15. Lambda Functions (One-Line Mini Functions)

A **lambda** is a short function written in a single line, without using `def`.

```python
lambda arguments: expression
```

**Real life example:** A lambda is like a quick sticky note instead of a full letter — use it only for small, one-time jobs.

````python
square_lambda = lambda x: x*x
print(square_lambda(7))

add_lambda = lambda a, b: a+b
print(add_lambda(4, 5))
````

## 16. Lambda with `map()` and `filter()`

- `map()` applies a function to every item in a list
- `filter()` keeps only items that satisfy a condition

````python
nums = [1, 2, 3, 4]

squares = list(map(lambda x: x*x, nums))
print(squares)

even = list(filter(lambda x: x%2==0, nums))
print(even)
````

## 17. Recursion (A Function Calling Itself)

A **recursive function** calls itself until it reaches a stopping condition (**base case**).

**Real life example:** Recursion is like **Russian nesting dolls** — each doll has a smaller one inside, until the smallest one that doesn't open (base case).

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))
````
